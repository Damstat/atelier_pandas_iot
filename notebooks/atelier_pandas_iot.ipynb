## 5) Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de mesure 12h, 13h, 14h et 15h

```python
# Méthode 1 : index passé lors de la création
temp_heures_1 = pd.Series([22.5, 24.1, 19.8, 26.3], index=["12h", "13h", "14h", "15h"])
temp_heures_1
```

```python
# Méthode 2 : création puis affectation de l'attribut .index
temp_heures_2 = pd.Series([22.5, 24.1, 19.8, 26.3])
temp_heures_2.index = ["12h", "13h", "14h", "15h"]
temp_heures_2
```

## 6) Afficher la 1ère et la dernière valeur

```python
print("Premier valeur",temp_heures_1.iloc[0])
print("Derniere valeur",temp_heures_1.iloc[-1])
```
